# CENSO ESCOLAR PREFEITURA DE FORTALEZA (2017-2021)

## Introdução

O censo escolar é um panorama de dados educacionais relacionados ao número de matriculados por região. No caso da prefeitura de Fortaleza, a classificação dessas regiões esta em seis distritos (https://educacao.sme.fortaleza.ce.gov.br/distritos-de-educacao/)

# IMPORTAÇÃO DE BIBLIOTECAS
## Link da base de dados



```python
import pandas as pd

url_2021 = "https://dados.fortaleza.ce.gov.br/dataset/4f3f42f9-80ca-4334-bc7d-9b7c69e2e191/resource/0836914f-b81a-4d95-b348-d2bce47e0169/download/matricula_2021.csv"
url_2020 = 'https://dados.fortaleza.ce.gov.br/dataset/c1634d0e-be7d-41ea-ac30-75c0b57760f4/resource/dd09ad31-af85-435e-bd71-1bb5d1b94959/download/matricula_2020.csv'
url_2019 = 'https://dados.fortaleza.ce.gov.br/dataset/4df4e191-ebfd-48bd-9a51-beeaf4d05ad4/resource/035c2e2d-46bb-4f05-9760-7f0fe67c4f42/download/matricula_2019.csv'
url_2018 = 'https://dados.fortaleza.ce.gov.br/dataset/5f1a4e39-d02c-4ec4-aeec-046137d65e3f/resource/2168e46c-943f-4521-84f1-14fd29e33256/download/matricula_2018.csv'
url_2017 = 'https://dados.fortaleza.ce.gov.br/dataset/2f68c9d2-9ba8-402b-b0a5-fafa5d9b914f/resource/51cdf5ed-7596-4243-9d33-6ea60b73ad90/download/matricula_2017.csv'
```

# INFORMAÇÃO DA BASE DE DADOS
## Verificação dos valores


*   Leitura do arquivo csv (https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html#pandas.read_csv);
* Reunião dos censos em um único dataframe;
*   Os 10 primeiros valores (https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html#pandas.DataFrame.head);
* Criação de uma coluna ano;
*   Obtenção dos valores nulos;

*   Informações gerais da base;
*   Estatística descritiva da base.







```python
# CENSO ESCOLAR 2021
censo_escolar_2021 = pd.read_csv(url_2021, sep=';')

print(censo_escolar_2021.head(5))
```

```python
# CENSO ESCOLAR 2020
censo_escolar_2020 = pd.read_csv(url_2020, sep=';')
print(censo_escolar_2020.head(5))
```

```python
# CENSO ESCOLAR 2019
censo_escolar_2019 = pd.read_csv(url_2019, sep=';')
print(censo_escolar_2019.head(5))
```

```python
# CENSO ESCOLAR 2018
censo_escolar_2018 = pd.read_csv(url_2018, sep=';', encoding="latin-1")
print(censo_escolar_2018.head(5))
```

```python
# CENSO ESCOLAR 2017
censo_escolar_2017 = pd.read_csv(url_2017, sep=';', encoding='latin-1')
print(censo_escolar_2017.head(5))
```

```python
# ADICIONAR A COLUNA ANO
censo_escolar_2021['ano'] = 2021
censo_escolar_2020['ano'] = 2020
censo_escolar_2019['ano'] = 2019
censo_escolar_2018['ano'] = 2018
censo_escolar_2017['ano'] = 2017
```

```python
# UNIÃO DO CONJUNTO DE DADOS
censo_escolar = pd.concat([censo_escolar_2021, censo_escolar_2020, censo_escolar_2019, censo_escolar_2018, censo_escolar_2017])
print(censo_escolar.head(5))
```

```python
# EXCLUIR COLUNA UNNAMED

censo_escolar = censo_escolar.drop(columns=['Unnamed: 16'])
print(censo_escolar.head(5))
```

```python
distritos = {"I":1, "II":2, "III":3, "IV":4, "V":5, "VI":6}
censo_escolar['distrito'] = censo_escolar['distrito'].astype(str).str.strip() # Tirar os espaços extras
censo_escolar['distrito'] = censo_escolar['distrito'].replace(distritos)

censo_escolar['distrito'] = pd.to_numeric(censo_escolar['distrito'], errors='coerce')
censo_escolar = censo_escolar.dropna(subset=['distrito']) # Remove linhas com valores nulos
censo_escolar['distrito'] = censo_escolar['distrito'].astype(int)
print(censo_escolar.head(5))
```

```python
# INFORMAÇÕES GERAIS DO DATAFRAME

censo_escolar.info()
```

```python
# ESTATÍSTICA DESCRITIVA

censo_escolar.describe()
```

O censo escolar apresenta as seguintes informações:


*   Divido por distritos de 1 à 6;
*   Tem um código do Instituto Nacional de Estudos e Pesquisas Escolares - INEP (https://www.gov.br/inep/pt-br) com a respectiva unidade de escola municipal;
*   Apresenta os alunos matriculados por distrito e escola: creche, pre_escola, 1_ano, 2_ano_ até 9_ano. Além disso, a Educação de Jovens e Adultos dos anos iniciais (ai) e anos finais (af);
*   Total de 1487 linhas X 18 colunas.

## Total de alunos matriculados (2017-2021)

```python
matriculados = ['creche', 'pre_escola', '1_ano', '2_ano', '3_ano',
                             '4_ano','5_ano', '6_ano', '7_ano', '8_ano', '9_ano',
                             'eja_ai', 'eja_af']

# SOMA CADA COLUNA E DEPOIS SOMA AS COLUNAS
total_alunos = censo_escolar[matriculados].sum().sum()
print(f"Total de alunos matriculados: \033[1m{int(total_alunos):,}\033[0m".replace(",","."))
```

Total de alunos matriculos na rede municipal de Fortaleza (2017-2021): 1.111.867.

O número em negrito correspondente ao método escape ANSI (https://www.delftstack.com/pt/howto/python/python-bold-text/?utm_source=copilot.com).


## Alunos matriculados em cada ano

```python
ano_matriculados = censo_escolar.groupby('ano',)[matriculados].sum().sum(axis=1)
ano_matriculados = ano_matriculados.apply(lambda x: f"{int(x):,}".replace(",", "."))
print(ano_matriculados)
```

O crescimento de alunos ao longo dos anos apresenta variação positiva, inclusive em período de pandemia de covid-19 (2020-2021).

## Taxa de variação de matriculados por ano

```python
matriculas_ano = censo_escolar.groupby('ano')[matriculados].sum()
matriculas_ano_taxa = matriculas_ano.pct_change() * 100
matriculas_ano_taxa = matriculas_ano_taxa.fillna(0)
matriculas_ano_taxa = matriculas_ano_taxa.round(2)
matriculas_ano_taxa = matriculas_ano_taxa.astype(str) + '%'
matriculas_ano_taxa
```



*   A educação considerada infantil (creche e pré-escola) tem aumento no índice de matriculados na pré-escola;
*   No ensino fundamental 1 (1 ano a 5 ano) tem uma taxa significativa do 1 ano e 2 ano, contudo no 5 ano apresenta queda;
*   Ao contrário do fundamental 1, o fundamental 2 apresenta mesma situação de queda, no entanto há um aumento percentual de matriculados no último ano do fundamental 2;
*   Já na Educação de Jovens e Adultos (EJA), a baixa adesão de matriculados tem regularidade durante os anos de amostra. Entende-se que há





## Alunos matriculados por distrito

```python
total_distrito = censo_escolar.groupby('distrito')[matriculados].sum().sum(axis=1)
total_distrito = total_distrito.apply(lambda x: f"{int(x):,}".replace(",", "."))
print(total_distrito)
```



*   O distrito com maiores números de matriculados é distrito 6 e, em seguida, o distrito 5. Localizados na região da grande.
*   Por outro lado, os distritos 1, 2 e 3 tem números abaixo e bem semelhantes de matriculados. São regiões.



## Desigualdade nos distritos

```python
media_distrito = censo_escolar.groupby(['distrito','ano'])[matriculados].sum().mean(axis=1)
desvio_distrito = censo_escolar.groupby(['distrito','ano'])[matriculados].sum().std(axis=1)

cft_var = (desvio_distrito/media_distrito) * 100
cft_var
```

## 10 escolas com maiores números de matrículas ##

```python
# TOP 10 ALUNOS MATRICULADOS POR UNIDADE ESCOLAR E DISTRITO

escolas = censo_escolar.groupby(['unidade_escolar', "distrito"])[matriculados].sum().sum(axis=1).sort_values(ascending=False)
top_10_escolas = escolas.head(10)
top_10_escolas = top_10_escolas.apply(lambda x: f"{int(x):,}".replace(",", "."))
print(top_10_escolas)
```

## Média de matriculados por distrito

```python
media_distrito = censo_escolar.groupby('distrito')[matriculados].sum().mean(axis=1)
media_distrito = media_distrito.apply(lambda x: f"{int(x):,}".replace(",", "."))
print(media_distrito)
```

## Média de alunos matriculados por ano

```python
media_ano = censo_escolar.groupby('ano')[matriculados].sum().mean(axis=1)
media_ano = media_ano.apply(lambda x: f"{int(x):,}".replace(",", "."))
print(media_ano)
```

## Quantidade de escolas por distrito

```python
quantidade_escolas = censo_escolar.groupby("distrito")["inep"].nunique()
quantidade_escolas
```

## Crescimento de matriculados por distrito

```python
matriculas_distrito = censo_escolar.groupby(['distrito','ano'])[matriculados].sum().sum(axis=1)
matriculas_distrito = matriculas_distrito.apply(lambda x: f"{int(x):,}".replace(",", "."))
matriculas_distrito
```

# VISUALIZAÇÃO DE DADOS



*   Gráfico de linha: alunos matriculados X ano;
*   Gráfico heatmap: variação de matrículas X ano;
*   Gráfico de barra: matriculados X distrito,  escolas X distrito;

## Gráfico de linha dos alunos matriculados por ano

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

ano_matriculados_numeric = censo_escolar.groupby('ano',)[matriculados].sum().sum(axis=1)

figure = plt.figure(figsize=(10, 6))
plt.plot(ano_matriculados_numeric.index, ano_matriculados_numeric, marker='o')
plt.title('Alunos matriculados por ano')
plt.xlabel('Ano')
plt.ylabel('Alunos matriculados')

# Força os ticks do eixo X a serem inteiros
plt.xticks(ano_matriculados_numeric.index.astype(int))

# Define os limites do eixo Y
plt.ylim(200000, 260000)

# Define intervalos fixos de 10.000
plt.yticks(range(200000, 260001, 10000))

# Formata o eixo Y com separador de milhar
plt.gca().yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{int(x):,}".replace(",", "."))
)

plt.show()
```

## Gráfico de mapa de calor da taxa variação de matrículas por ano

```python
import seaborn as sns

matriculas_ano_taxa_num = matriculas_ano_taxa.apply(lambda x: x.str.rstrip('%').astype(float))
plt.figure(figsize=(12,6))
sns.heatmap(matriculas_ano_taxa_num, annot=True, fmt=".1f", cmap="RdYlGn", center=0)
plt.title("Mapa de calor da taxa de variação por nível")
plt.show()


```

## Gráfico de barra dos alunos matriculados por distrito

```python
figure = plt.figure(figsize=(10, 6))
plt.bar(total_distrito.index, total_distrito, width=0.5)
plt.title('Alunos matriculados por distrito')
plt.xlabel('Distrito')
plt.ylabel('Alunos matriculados')
plt.grid(axis='y', linestyle='--', alpha=0.9)
plt.show()
```

## Gráfico de barra da quantidade de escolas por distrito

```python
figure = plt.figure(figsize=(10, 6))
plt.bar(quantidade_escolas.index, quantidade_escolas, width=0.5)
plt.title('Quantidade de escolas por distrito')
plt.xlabel('Distrito')
plt.ylabel('Quantidade de escolas')
plt.show()
```

## Gráfico de série temporal do crescimento de matrículas por distrito

```python
for d in matriculas_distrito.index.levels[0]:
    dados = matriculas_distrito.loc[d].str.replace('.', '').astype(int)
    plt.plot(dados.index, dados, marker='o', label=f'Distrito {d}')

plt.title("Crescimento das matrículas por distrito")
plt.xlabel("Ano")
plt.ylabel("Matrículas")


plt.legend(
    title="Distritos",
    bbox_to_anchor=(1, 1),   # posiciona fora do gráfico, à direita
    loc='upper left',           # canto superior esquerdo da caixa
    fontsize=8,                # tamanho da fonte
    ncol=2,                     # duas colunas
    frameon=True,               # com borda
    facecolor='white',          # fundo branco
    framealpha=0.8              # leve transparência
)

plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```
